```python
import pandas as pd
import numpy as np
import sklearn
import hashlib
import matplotlib
print(pd.__version__, np.__version__, sklearn.__version__, matplotlib.__version__)
```

```python

key_df = pd.read_csv('key.csv', dtype={'pos': int, 'shift': str, 'relay': str})
message_df = pd.read_csv('message.csv')
codebook_df = pd.read_csv('codebook.csv')
key_df['shift'] = pd.to_numeric(key_df['shift'], errors='coerce')
N = len(codebook_df)
key_df.loc[key_df['shift'] >= N, 'shift'] = np.nan
##print(key_df.groupby('relay')['shift'].count())
##print(key_df.groupby('relay')['shift'].nunique())
##print(key_df[key_df['relay'] == 'MontDesCats-77']['shift'])
##print(key_df[key_df['relay'] == 'Orval-07']['shift'])
##before = key_df.groupby('relay')['shift'].count()
##key_df['shift'] = pd.to_numeric(key_df['shift'], errors ='coerce')
##after = key_df.groupby('relay')['shift'].count()
##print(before-after)
bad_stations_1_4 = ['Spencer-61', 'MontDesCats-77', 'Orval-07', 'Achel-19']

remaining_df = key_df[~key_df['relay'].isin(bad_stations_1_4)]
pivot = remaining_df.pivot(index='pos', columns='relay', values='shift')
approx_key = pivot.mode(axis=1).iloc[:, 0]
mismatches = {}
for col in pivot.columns:
    station_data = pivot[col]
    valid_mask = station_data.notna()
    errors = (station_data[valid_mask] != approx_key[valid_mask]).sum()
    error_rate = errors / valid_mask.sum()
    mismatches[col] = error_rate
mismatch_series = pd.Series(mismatches).sort_values(ascending=False)
deceptive_stations = mismatch_series.index[:2].tolist()

print(mismatch_series)


honest_df = remaining_df[~remaining_df['relay'].isin(deceptive_stations)]

honest_pivot = honest_df.pivot(index='pos', columns='relay', values='shift')
final_key = honest_pivot.mode(axis=1).iloc[:, 0]

key_sum = int(final_key.sum())
print(key_sum)

key_list_str = final_key.astype(int).astype(str).tolist()
key_hash = hashlib.sha256(",".join(key_list_str).encode()).hexdigest()

```
